```python
import pandas as pd
import os
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import acf
import numpy as np
import sys
sys.path.append(os.path.dirname(os.path.dirname(os.getcwd())))
from data_provider.m4 import M4Dataset, M4Meta

import matplotlib
matplotlib.rcParams['pdf.fonttype'] = 42
matplotlib.rcParams['ps.fonttype'] = 42
```

```python
USRDIR = '/mnt/tidalfs-bdsz01/usr/panlicheng'
df_raw = pd.read_csv(f'{USRDIR}/dataset/ETT-small/ETTh1.csv')
scaler = StandardScaler()
cols_data = df_raw.columns[1:]
df_data = df_raw[cols_data]

train_data = df_data[:12*30*24]
scaler.fit(train_data.values)
data = scaler.transform(df_data.values)

train_data = data[:12*30*24]
test_data = data[16*30*24:20*30*24]


acf_values = acf(train_data[:,0], nlags=200)


lags = np.arange(len(acf_values))
plt.figure(figsize=(3, 2.5))
plt.bar(lags, acf_values)

plt.xlim([0, 200])
plt.ylim([0, 1])

plt.xlabel("Lags", fontsize=10)
plt.ylabel("Autocorrelation", fontsize=10)

plt.grid(True)
# plt.savefig('acf/etth1[0].pdf', bbox_inches='tight')
# plt.show()
```

```python
USRDIR = '/mnt/tidalfs-bdsz01/usr/panlicheng'
dataset = M4Dataset.load(training=True, dataset_file=f'{USRDIR}/dataset/m4')
seasonal_patterns = "Yearly"
seasonal_patterns = "Quarterly"
seasonal_patterns = "Monthly"
seasonal_patterns = "Weekly"
seasonal_patterns = "Daily"
seasonal_patterns = "Hourly"

training_values = [
    np.array(v[~np.isnan(v)]) 
    for v in dataset.values[dataset.groups == seasonal_patterns]
]
ids = np.array([i for i in dataset.ids[dataset.groups == seasonal_patterns]])
# print(ids)
timeseries = [ts for ts in training_values]
print(max([len(ts) for ts in timeseries]))
print(min([len(ts) for ts in timeseries]))
df_data = np.concatenate(timeseries, axis=0)
print(df_data.shape)

if seasonal_patterns == "Weekly":
    acf_values = acf(df_data, nlags=10000)
elif seasonal_patterns == "Daily":
    acf_values = acf(df_data, nlags=20000)
else:
    acf_values = acf(df_data, nlags=2000)
lags = np.arange(len(acf_values))
# plt.figure(figsize=(10, 2.5))
# plt.plot(df_data)

plt.figure(figsize=(10, 2.5))
plt.bar(lags, acf_values)

if seasonal_patterns == "Yearly":
    plt.xlim([1000, 1100])
    plt.ylim([0, 0.2])
elif seasonal_patterns == "Quarterly":
    plt.xlim([1400, 1800])
    plt.ylim([0, 0.2])
elif seasonal_patterns == "Monthly":
    plt.xlim([1000, 1200])
    plt.ylim([0, 0.2])
elif seasonal_patterns == "Weekly":
    plt.xlim([2000, 6000])
    plt.ylim([-0.2, 0.2])
elif seasonal_patterns == "Daily":
    plt.xlim([2500, 10000])
    plt.ylim([-1, 1])
elif seasonal_patterns == "Hourly":
    plt.xlim([1000, 1100])
    plt.ylim([0, 0.1])

plt.xlabel("Lags", fontsize=10)
plt.ylabel("Autocorrelation", fontsize=10)

plt.grid(True)
```
