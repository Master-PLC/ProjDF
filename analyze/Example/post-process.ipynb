# Remove folders with low mse

```python
import os
import shutil
import numpy as np

root_dir = '/mnt/tidalfs-bdsz01/usr/panlicheng/workspace/ProjDF-Meta/results_ML3/finetune'

dst_bar = {
    # 'ETTm1': 0.6,
    # 'ETTm2': 0.55,
    # 'ETTh1': 0.55,
    # 'ETTh2': 0.5,
    # 'ECL': 0.3,
    # 'Traffic': 
    # 'Weather': 0.42,
    'PEMS03': 0.2,
    'PEMS08': 0.22
}

for dst, bar in dst_bar.items():
    count = 0
    for root, dirs, files in os.walk(root_dir):
        if not dst in root:
            continue
        for file in files:
            if file == 'metrics.npy':
                metrics_path = os.path.join(root, file)
                metrics = np.load(metrics_path)
                mae, mse = metrics[0], metrics[1]
                if mse > bar:
                    base_root = os.path.dirname(os.path.dirname(root))
                    # print(base_root)
                    shutil.rmtree(base_root)
                    count += 1
    print(f'{dst}: removed {count} folders')
```

# Tar single folder

```python
#!/usr/bin/env python3
import os
import tarfile
import glob
from pathlib import Path

def archive_single_folder(folder_path):
    """处理单个文件夹，打包指定文件类型"""
    folder_path = Path(folder_path)
    if not folder_path.exists():
        print(f"Error: Folder {folder_path} does not exist")
        return False
    
    # 定义要打包的文件模式
    patterns = ['*.txt', '*.log', '*.pdf', 'events*', '*.pth']
    
    # 收集所有匹配的文件
    files_to_archive = []
    for pattern in patterns:
        files_to_archive.extend(folder_path.rglob(pattern))
    print(files_to_archive)
    
    if not files_to_archive:
        print(f"No files to archive in {folder_path}")
        return False
    
    # 创建压缩文件
    archive_path = folder_path / 'archive.tar.xz'
    
    with tarfile.open(archive_path, 'w:xz') as tar:
        for file_path in files_to_archive:
            # 计算相对路径
            arcname = file_path.relative_to(folder_path)
            tar.add(file_path, arcname=arcname)
    
    print(f"Archived {len(files_to_archive)} files to {archive_path}")
    
    # 删除原文件
    for file_path in files_to_archive:
        file_path.unlink()
    
    print(f"Deleted {len(files_to_archive)} original files")
    return True

# 使用示例
target_folder = "/mnt/tidalfs-bdsz01/usr/panlicheng/workspace/ProjDF-Meta/results_ML3/finetune/Fredformer_ECL_96_1.0_0.0_0.002_0.002_0.1_type1_100_5_32_64_0.0_5.0_3_0.0_1_MSE_1_300"
archive_single_folder(target_folder)
```

# untar single folder

```python
#!/usr/bin/env python3
import tarfile
from pathlib import Path

def extract_single_folder(folder_path):
    """解压单个文件夹中的archive.tar.xz"""
    folder_path = Path(folder_path)
    archive_path = folder_path / 'archive.tar.xz'
    
    if not archive_path.exists():
        print(f"Error: Archive {archive_path} does not exist")
        return False
    
    try:
        with tarfile.open(archive_path, 'r:xz') as tar:
            # 解压到当前文件夹
            tar.extractall(path=folder_path)
            # 获取解压的文件列表
            members = tar.getnames()
        
        print(f"Extracted {len(members)} files from {archive_path}")
        
        # 可选：删除压缩包
        archive_path.unlink()
        print(f"Deleted archive {archive_path}")
        
        return True
    
    except Exception as e:
        print(f"Error extracting {archive_path}: {e}")
        return False

# 使用示例
target_folder = "/mnt/tidalfs-bdsz01/usr/panlicheng/workspace/ProjDF-Meta/results_ML3/finetune/Fredformer_ECL_96_1.0_0.0_0.002_0.002_0.1_type1_100_5_32_64_0.0_5.0_3_0.0_1_MSE_1_300"
extract_single_folder(target_folder)
```

# Tar total folder

```python
#!/usr/bin/env python3
import os
import tarfile
import glob
from pathlib import Path
import sys

def archive_files_in_folder(folder_path):
    """在指定文件夹中打包并删除匹配的文件"""
    folder_path = Path(folder_path)
    
    # 定义要打包的文件模式
    patterns = ['*.txt', '*.log', '*.pdf', 'events*', '*.pth']
    
    # 收集所有匹配的文件
    files_to_archive = []
    for pattern in patterns:
        files_to_archive.extend(folder_path.rglob(pattern))
    
    if not files_to_archive:
        return 0
    
    # 创建压缩文件
    archive_path = folder_path / 'archive.tar.xz'
    
    try:
        with tarfile.open(archive_path, 'w:xz') as tar:
            for file_path in files_to_archive:
                arcname = file_path.relative_to(folder_path)
                tar.add(file_path, arcname=arcname)
        
        # 删除原文件
        for file_path in files_to_archive:
            file_path.unlink()
        
        return len(files_to_archive)
    
    except Exception as e:
        print(f"Error processing {folder_path}: {e}")
        return -1

def process_all_folders(base_path="."):
    """处理指定目录下的所有文件夹"""
    base_path = Path(base_path)
    
    if not base_path.exists():
        print(f"Error: Base path {base_path} does not exist")
        return
    
    # 获取所有子文件夹（只处理直接子文件夹）
    subfolders = [f for f in base_path.iterdir() if f.is_dir()]
    
    total_processed = 0
    total_files = 0
    
    for folder in subfolders:
        files_count = archive_files_in_folder(folder)
        if files_count > 0:
            print(f"✓ {folder.name}: {files_count} files archived")
            total_processed += 1
            total_files += files_count
        elif files_count == 0:
            continue  # 静默跳过没有匹配文件的文件夹
        else:
            print(f"✗ {folder.name}: Error occurred")
    
    print(f"\nSummary: {total_processed} folders processed, {total_files} files archived")

# 可以通过命令行参数指定基础路径，默认为当前目录
base_path = '/mnt/tidalfs-bdsz01/usr/panlicheng/workspace/ProjDF-Meta/results_ML3/finetune'
process_all_folders(base_path)
```
